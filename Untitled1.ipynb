```python

# Biased Quantum Coin Experiment using Qiskit

# 1. Import libraries
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 2. Initialize simulator
simulator = AerSimulator()
shots = 100

# 3. Circuit A: Make outcome 0 more likely
circuit_a = QuantumCircuit(1, 1)
circuit_a.ry(0.6, 0)       # Rotation around Y axis
circuit_a.measure(0, 0)

# 4. Circuit B: Make outcome 1 more likely
circuit_b = QuantumCircuit(1, 1)
circuit_b.ry(2.4, 0)       # Larger rotation around Y axis
circuit_b.measure(0, 0)

# 5. Display circuits
print("Circuit A: Outcome 0 is more likely")
display(circuit_a.draw("mpl"))

print("Circuit B: Outcome 1 is more likely")
display(circuit_b.draw("mpl"))

# 6. Predictions before execution
print("\nPREDICTIONS")
print("Circuit A: About 91% outcome 0 and 9% outcome 1")
print("Circuit B: About 13% outcome 0 and 87% outcome 1")

# 7. Execute circuits for 100 shots
circuits = [circuit_a, circuit_b]
names = ["Circuit A (0 more likely)", "Circuit B (1 more likely)"]
results = {}

print("\nMEASUREMENT RESULTS")

for name, circuit in zip(names, circuits):
    job = simulator.run(circuit, shots=shots)
    result = job.result()
    counts = result.get_counts()
    results[name] = counts

    zeros = counts.get("0", 0)
    ones = counts.get("1", 0)

    print(f"\n{name}")
    print("Counts:", counts)
    print(f"Outcome 0: {zeros}/100 = {zeros}%")
    print(f"Outcome 1: {ones}/100 = {ones}%")

# 8. Display histograms
for name, counts in results.items():
    plot_histogram(counts, title=name)
    plt.show()

# 9. Short explanation
print("\nEXPLANATION")
print("The Ry gate rotates the qubit state by a chosen angle.")
print("Changing the angle changes the probabilities of measuring 0 or 1.")
print("A smaller angle favors 0; a larger angle can favor 1.")

# 10. Challenge question answer
print("\nCHALLENGE QUESTION")
print(
    "Quantum gates change the qubit's state and its probability amplitudes. "
    "When measured, the probabilities of 0 and 1 are determined by the "
    "squared magnitudes of those amplitudes. By changing the gate type "
    "or rotation angle, we can make either outcome more likely."
)
```
